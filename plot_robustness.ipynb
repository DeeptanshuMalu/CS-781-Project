```python
import matplotlib.pyplot as plt
import os
import numpy as np
```

```python
num_cars_0_path = "data/test/0/robustness/"
num_cars_1_path = "data/test/1/robustness/"
```

```python
max_iterations = 15
num_iterations = min(len(os.listdir(num_cars_0_path)), max_iterations)
```

```python
robustness_0 = []
accuracy_0 = []
for iteration_num in range(1, num_iterations + 1):
	with open(num_cars_0_path + f"robustness_metrics_iteration_{iteration_num}.txt", 'r') as f:
		lines = f.readlines()
		robustness_0.append(float(lines[0].strip()))
		accuracy_0.append(float(lines[1].strip()))
```

```python
robustness_1 = []
accuracy_1 = []
for iteration_num in range(1, num_iterations + 1):
	with open(num_cars_1_path + f"robustness_metrics_iteration_{iteration_num}.txt", 'r') as f:
		lines = f.readlines()
		robustness_1.append(float(lines[0].strip()))
		accuracy_1.append(float(lines[1].strip()))
```

```python
os.makedirs('experiments', exist_ok=True)

fig, ax = plt.subplots(1, 1, figsize=(8, 5))

ax.plot(robustness_0, label='0 cars')
ax.plot(robustness_1, label='1 car')
ax.plot((np.array(robustness_0)+np.array(robustness_1))/2, label='Avg Robustness')
ax.set_xticks(range(len(robustness_0)))
ax.set_ylabel('Robustness')
ax.set_xlabel('Iteration Number')
# ax.plot(robustness_0_a, label='0 cars normal', linestyle='--')
# ax.plot(robustness_1_a, label='1 car normal', linestyle='--')
# ax.plot((np.array(robustness_0_a)+np.array(robustness_1_a))/2, label='Avg Robustness normal', linestyle='--')
# ax.legend()
fig.savefig('experiments/robustness_plot.png')

handles, labels = ax.get_legend_handles_labels()
fig_legend = plt.figure()
fig_legend.legend(handles, labels, loc="center", ncol=len(labels))
fig_legend.savefig(f"experiments/legend_robustness.png", bbox_inches="tight")

fig, ax = plt.subplots(1, 1, figsize=(8, 5))

ax.plot(accuracy_0, label='0 cars')
ax.plot(accuracy_1, label='1 car')
ax.plot((np.array(accuracy_0)+np.array(accuracy_1))/2, label='Avg Accuracy')
ax.set_xticks(range(len(accuracy_0)))
ax.set_ylabel('Accuracy')
ax.set_xlabel('Iteration Number')
# ax.plot(accuracy_0_a, label='0 cars normal', linestyle='--')
# ax.plot(accuracy_1_a, label='1 car normal', linestyle='--')
# ax.plot((np.array(accuracy_0_a)+np.array(accuracy_1_a))/2, label='Avg Accuracy normal', linestyle='--')
# ax.legend()
fig.savefig('experiments/accuracy_plot.png')

handles, labels = ax.get_legend_handles_labels()
fig_legend = plt.figure()
fig_legend.legend(handles, labels, loc="center", ncol=len(labels))
fig_legend.savefig(f"experiments/legend_accuracy.png", bbox_inches="tight")
```
